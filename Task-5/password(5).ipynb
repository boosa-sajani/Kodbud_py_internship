```python
import re

while True:
    password = input("ENTER YOUR PASSWORD: ")

    length = len(password) >= 8
    num = re.search(r"\d", password)
    upper = re.search(r"[A-Z]", password)
    special = re.search(r"[!@#$%^&*(),.?\":{}|<>]", password)

    if length and num and upper and special:
        print("STRONG PASSWORD!!")
        confirm = input("CONFIRM YOUR PASSWORD: ")
        if password == confirm:
            print(" PASSWORD CONFIRMED SUCCESSFULLY!!")
            break
        else:
            print("Password does'nt matched. Try again.\n")

    else:
        print("WEAK PASSWORD")

        print("\n SUGGESTIONS")
        if not length:
            print("- Use at least 8 characters")
        if not num:
            print("- Add at least one number (0-9)")
        if not upper:
            print("- Add at least one uppercase letter (A-Z)")
        if not special:
            print("- Add at least one special character (!@#$...)")

        print("\n PLEASE TRY AGAIN.\n")
```
