```python
n1=float(input("ENTER NUMBER 1:"))
n2=float(input("ENTER NUMBER 2:"))
print("\n SELECT OPERATION:")
print("1.ADDITION(+)")
print("2.SUBTRACION(-)")
print("3.MULTIPLICATION(*)")
print("4.DIVISON(/)")
print("5.MODULO(%)")
choice=input("ENTER CHOICE (1/2/3/4/5):")
if choice=='1':
    result=n1+n2
    print("RESULT:",result)
elif choice=='2':
    result=n1-n2
    print("RESULT:",result)
elif choice=='3':
    result=n1*n2
    print("RESULT:",result)
elif choice=='4':
    if n2!=0:
        result=n1/n2
        print("RESULT:",result)
    else:
        print("ERROR: DIVISION BY ZERO IS NOT ALLOWED")
elif choice=='5':
    result=n1%n2
    print("RESULT",result)
else:
    print("INVALID CHOICE")
```
